### How to download data from Google Drive

```python
import gdown 
```

```python
url = "https://drive.google.com/file/d/1NjnvMGMwJr94Q64BzkdA_MBG_IIjPdFB/view?usp=sharing"
```

```python
type(url)
```

```python
file_id = url.split("/")[-2]
file_id
```

```python
prefix = 'https://drive.google.com/uc?/export=download&id='
gdown.download(prefix+file_id, "Chest-CT-Scan-data.zip")
```
